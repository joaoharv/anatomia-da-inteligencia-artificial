## Introdução

Historicamente, o processamento de texto dependia de redes recorrentes (RNNs e LSTMs), que processavam sequências palavra por palavra, sofrendo com o problema do esquecimento de longo prazo e gargalos de paralelização.

LLMs são sofisticados sistemas de Inteligência Artificial, especificamente desenvolvidos para processar e compreender a linguagem natural humana em grande escala. Utilizando técnicas avançadas de aprendizado de máquina, como redes neurais profundas, LLMs são treinados em grandes volumes de dados textuais para aprender padrões, contextos semânticos e nuances da linguagem. Esses modelos podem ser empregados em diversas tarefas, como geração de texto, tradução automática, resumo de documentos e interações conversacionais, destacando-se pela capacidade de realizar essas tarefas com uma notável fluência e contextualização.

O advento do artigo [*Attention Is All You Need*](https://arxiv.org/abs/1706.03762) (Vaswani et al., 2017) introduziu o Transformer, cuja espinha dorsal é o mecanismo de ***Self-Attention*** (Atenção Autodirigida).

> **Attention:** Permite que o modelo pondere a importância de diferentes palavras em uma frase, independentemente da distância entre elas. Para a palavra "banco", o modelo usa o contexto para decidir se o significado é financeiro ou um móvel de praça.

> **Multi-Head Attention:** O mecanismo é dividido em múltiplos "cabeçotes" paralelos, permitindo que a rede preste atenção simultaneamente em diferentes aspectos sintáticos e semânticos (ex: sujeito, tempo verbal, concordância).

Aqui estão alguns exemplos específicos de LLMs em uso hoje:
* O **LaMDA**, da Google, é um LLM que pode gerar texto criativo, como poemas, histórias, scripts, etc.
* O **GPT-3**, da OpenAI, é um LLM que pode ser usado para responder a perguntas de forma informativa, mesmo que sejam abertas.
* O **LLaMA**, da Meta, é um LLM que pode ser usado para traduzir texto de um idioma para outro.

## Tokenização e Embeddings
Os computadores não leem palavras diretamente; o texto passa por etapas fundamentais antes de entrar na rede neural:

* **Tokenização por Subpalavras (*Subword Tokenization*):** Algoritmos como *Byte-Pair Encoding* (BPE) ou *WordPiece* dividem palavras complexas ou raras em pedaços menores (tokens). Isso resolve o problema de vocabulário desconhecido (*Out-of-Vocabulary*).
* **Embeddings:** Cada token é mapeado para um vetor denso de alta dimensionalidade em um espaço contínuo, onde palavras semanticamente similares ficam próximas.
*  ***Positional Encoding*:** Como o Transformer processa todos os tokens em paralelo (perseverando a velocidade), ele perde a noção de ordem temporal. Vetores de codificação posicional são somados aos embeddings para injetar a posição exata de cada token na frase.


## Paradigmas de Treinamento

Os LLMs modernos passam por um ciclo de vida estruturado em etapas:

1. **Pré-treinamento não supervisionado:** O modelo consome bilhões de páginas de texto com tarefas auto-supervisionadas (ex: prever a próxima palavra ou preencher lacunas), aprendendo gramática, fatos sobre o mundo e lógica básica.

2. ***Fine-Tuning* Tradicional**: Ajuste de todos ou parte dos pesos do modelo pré-treinado em uma base de dados específica para uma tarefa a jusante (downstream task), como análise de sentimentos ou resumo.

3. ***In-Context Learning* e *Prompt Engineering*:** Com modelos muito grandes, muitas vezes não é necessário alterar os pesos. Fornecer instruções e exemplos dentro do próprio prompt (Few-shot prompting) guia o comportamento do modelo instantaneamente.

## Reconhecimento de Entidades

O reconhecimento de entidade nomeada (NER) é uma tarefa em que o modelo precisa descobrir quais partes do texto de entrada correspondem a entidades, como pessoas, locais ou organizações.

```python
# Para esta tarefa, vamos utilizar Tokenizers a fim de preparar os dados de entrada.
from transformers import BertForTokenClassification, DistilBertTokenizerFast
```

```python
model = BertForTokenClassification.from_pretrained('monilouise/ner_pt_br')
tokenizer = DistilBertTokenizerFast.from_pretrained('neuralmind/bert-base-portuguese-cased', model_max_length=512, do_lower_case=False)
ner = pipeline('ner', model=model, tokenizer=tokenizer, aggregation_strategy="simple")
```

```python
ner("Meu nome é João e eu trabalho na agência do Banco do Brasil dentro Rede Globo em São Paulo.")
```

## Geração de Resposta (Q&A)

Este pipeline responde a perguntas usando informações de um determinado contexto.

```python
text_generator = pipeline("text-generation", model="Qwen/Qwen2.5-1.5B-Instruct", device_map="auto")
```

```python
# Definimos a pergunta e o contexto
question = "Where do I work?"
context = "My name is John and I have been working at Facebook in California"

# Estruturamos as mensagens no formato que o modelo de chat espera
messages = [
    {"role": "system", "content": "You are a helpful assistant. Answer the question based ONLY on the provided context."},
    {"role": "user", "content": f"Context: {context}\n\nQuestion: {question}"}
]

# Geramos a resposta
outputs = text_generator(messages, max_new_tokens=50)
print(outputs[0]["generated_text"][-1]["content"])
```

## Sumarização de Textos

A sumarização é a tarefa de reduzir um texto a um texto mais curto, mantendo todos (ou a maioria) dos aspectos importantes referenciados no texto.

```python
def summarize_text(text):
    messages = [
        {"role": "system", "content": "Você é um assistente útil. Resuma o texto fornecido de forma clara e concisa."},
        {"role": "user", "content": f"Texto para resumir:\n{text}"}
    ]
    outputs = text_generator(messages, max_new_tokens=150)
    return outputs[0]["generated_text"][-1]["content"]
```

```python
text_to_summarize = """
    America has changed dramatically during recent years. Not only has the number of
    graduates in traditional engineering disciplines such as mechanical, civil,
    electrical, chemical, and aeronautical engineering declined, but in most of
    the premier American universities engineering curricula now concentrate on
    and encourage largely the study of engineering science. As a result, there
    are declining offerings in engineering subjects dealing with infrastructure,
    the environment, and related issues, and greater concentration on high
    technology subjects, largely supporting increasingly complex scientific
    developments. While the latter is important, it should not be at the expense
    of more traditional engineering.

    Rapidly developing economies such as China and India, as well as other
    industrial countries in Europe and Asia, continue to encourage and advance
    the teaching of engineering. Both China and India, respectively, graduate
    six and eight times as many traditional engineers as does the United States.
    Other industrial countries at minimum maintain their output, while America
    suffers an increasingly serious decline in the number of engineering graduates
    and a lack of well-educated engineers.
"""

resumo = summarize_text(text_to_summarize)
print(resumo)
```

## Tradução Automática

A tradução automática refere-se ao uso de algoritmos e modelos de linguagem computacional para traduzir automaticamente o texto de uma língua para outra.

```python
def translate_text(text):
    messages = [
        {"role": "system", "content": "Você é um tradutor profissional. Traduza o texto a seguir do inglês para o português de forma natural."},
        {"role": "user", "content": text}
    ]
    outputs = text_generator(messages, max_new_tokens=256)
    return outputs[0]["generated_text"][-1]["content"]
```

```python
texto_ingles = """
            In the annals of British monarchy, the multifaceted tapestry of
            regal history unfolds with intricate nuances, as sovereigns and
            consorts alike navigate the labyrinthine corridors of power."""

traducao = translate_text(texto_ingles)
print(traducao)
```

10. Desenvolva uma função de tradução automatizada (`translate_text`) estruturada com mensagens de chat que configure o modelo como um tradutor profissional sênior. Submeta um parágrafo técnico complexo da área de Enfermagem do inglês para o português, garantindo fluidez natural e preservação de termos técnicos.

```python
# Exercício 10: tradução automática (inglês -> português) com persona de tradutor sênior
def translate_text(text):
    messages = [
        {"role": "system", "content": (
            "Você é um tradutor profissional sênior, especialista na área de Enfermagem e saúde. "
            "Traduza o texto a seguir do inglês para o português do Brasil de forma fluida e natural, "
            "preservando a precisão e os termos técnicos da área (por exemplo, nomes de procedimentos, "
            "medicamentos e siglas). Responda apenas com a tradução."
        )},
        {"role": "user", "content": text},
    ]
    outputs = text_generator(messages, max_new_tokens=300, do_sample=False)
    return outputs[0]["generated_text"][-1]["content"]


paragrafo_enfermagem = """
    Prior to administering intravenous medication, the nurse must verify the patient's
    identity using two identifiers, review the medication administration record and
    assess the patient for allergies. After inserting the peripheral venous catheter
    under aseptic technique, the nurse should monitor vital signs, observe the
    insertion site for signs of phlebitis or infiltration, and document any adverse
    reaction in the electronic health record.
"""

print(translate_text(paragrafo_enfermagem))
```
